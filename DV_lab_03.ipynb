```python
import csv

processed_students = []

# We will read 'students_marks.csv' which was uploaded as 'students_marks - students_marks.csv'
# Let's inspect which file exists and use the appropriate filename
import os
csv_filename = 'students_marks.csv'
if not os.path.exists(csv_filename):
    if os.path.exists('students_marks - students_marks.csv'):
        csv_filename = 'students_marks - students_marks.csv'

with open(csv_filename, mode='r', encoding='utf-8') as file:
    reader = csv.DictReader(file)

    # The subjects are Math, Stats, Python, Databases, English
    subject_fields = ['Math', 'Stats', 'Python', 'Databases', 'English']

    for row in reader:
        total_marks = 0.0
        count = 0

        # Convert the five mark fields to numbers
        for subject in subject_fields:
            if subject in row and row[subject]:
                val = float(row[subject])
                row[subject] = val
                total_marks += val
                count += 1

        # Calculate and round average to 2 decimal places
        if count > 0:
            row['Average'] = round(total_marks / count, 2)
        else:
            row['Average'] = 0.0

        processed_students.append(row)

# Print the first few students to verify the result
print(f"Processed {len(processed_students)} student records. Sample data:")
for student in processed_students[:5]:
    print(student)
```

```python
section_counts = {}

for student in processed_students:
    section = student.get('Section')
    if section:
        section_counts[section] = section_counts.get(section, 0) + 1

print("Students count in each section:")
for section, count in sorted(section_counts.items()):
    print(f"Section {section}: {count} students")
```

```python
#__________________ Task # 03 :Dates _________________#

import json
from datetime import datetime, date

# 1. Turn every Enrolled text into a real date with strptime
for student in info:
    # Raw date is in DD/MM/YYYY format based on info[0]['Enrolled'] -> "02/09/2024"
    student['Enrolled_date'] = datetime.strptime(student['Enrolled'], "%d/%m/%Y").date()

# 2. Count how many students joined in each intake (using strftime("%Y-%m"))
intake_counts = {}
for student in info:
    intake_key = student['Enrolled_date'].strftime("%Y-%m")
    intake_counts[intake_key] = intake_counts.get(intake_key, 0) + 1

print("Students joined in each intake (Year-Month):")
for intake, count in sorted(intake_counts.items()):
    print(f"{intake}: {count} students")

# 3. Print the number of days the first student has been enrolled on 1 November 2026
first_student_date = info[0]['Enrolled_date']
target_date = date(2026, 11, 1)
days_enrolled = (target_date - first_student_date).days
print(f"\nFirst student ({info[0]['Name']}) enrolled date: {first_student_date.isoformat()}")
print(f"Days enrolled on 1 Nov 2026: {days_enrolled}")

# 4. Save Name and Enrolled date for all students to enrolled.json with a custom DateEncoder
enrolled_list = [
    {"Name": s["Name"], "Enrolled": s["Enrolled_date"]}
    for s in info
]

class DateEncoder(json.JSONEncoder):
    def default(self, obj):
        if isinstance(obj, (date, datetime)):
            return obj.isoformat()
        return super().default(obj)

with open("enrolled.json", "w", encoding='utf-8') as f:
    json.dump(enrolled_list, f, indent=2, cls=DateEncoder)

print(f"\nSuccessfully saved student records to enrolled.json")
```

```python
#_______________________Task # 04 : SQL _______________________#

from sqlalchemy import create_engine, Column, Integer, String, Float
from sqlalchemy.orm import declarative_base, sessionmaker

# 1. Create lab3.db with a table students (id, Name, Section, Python) using SQLAlchemy
engine = create_engine("sqlite:///lab3.db")
Base = declarative_base()

class Student(Base):
    __tablename__ = "students"
    id = Column(Integer, primary_key=True, autoincrement=True)
    Name = Column(String)
    Section = Column(String)
    Python = Column(Float)

    def __repr__(self):
        return f"<Student(Name={self.Name}, Section={self.Section}, Python={self.Python})>"

# Start fresh and create the table
Base.metadata.drop_all(engine)
Base.metadata.create_all(engine)

# Open database session
Session = sessionmaker(bind=engine)
session = Session()

# Populate table using processed_students list
students_to_add = [
    Student(
        Name=s["Name"],
        Section=s.get("Section", ""),
        Python=float(s["Python"])
    )
    for s in processed_students
]

session.add_all(students_to_add)
session.commit()

# 2. Print the number of rows
row_count = session.query(Student).count()
print(f"Number of rows added to 'students' table: {row_count}\n")

# 3. Print the Section B students with Python 70 or more, highest first
query_results = (
    session.query(Student)
    .filter(Student.Section == "B", Student.Python >= 70)
    .order_by(Student.Python.desc())
    .all()
)

print("Section B students with Python >= 70 (highest first):")
for s in query_results:
    print(f"Name: {s.Name:<20} | Section: {s.Section} | Python: {s.Python}")
```

```python
from sqlalchemy import create_engine, text
from datetime import datetime
import os

# 1. Determine correct database path
db_filename = 'university.db'
if not os.path.exists(db_filename):
    if os.path.exists('university (1).db'):
        db_filename = 'university (1).db'

# Map RollNo to Student Name using our existing metadata
student_name_map = {s["RollNo"]: s["Name"] for s in info}

engine = create_engine(f"sqlite:///{db_filename}")

with engine.connect() as conn:
    print(f"--- Analyzing Fees from '{db_filename}' ---")

    # Select necessary columns based on actual schema: ['RollNo', 'Semester', 'Amount', 'DueDate', 'PaidOn']
    query = text("SELECT RollNo, DueDate, PaidOn, Amount FROM fees")
    records = conn.execute(query).fetchall()

    late_students = []
    unpaid_students = []
    total_collected = 0.0

    for row in records:
        roll_no = row[0]
        due_str = row[1]
        paid_str = row[2]
        amount = row[3]

        # Retrieve student name from our mapping using RollNo
        name = student_name_map.get(roll_no, roll_no)

        # Check if paid or unpaid
        if paid_str and str(paid_str).strip().upper() != "NULL" and str(paid_str).strip() != "":
            due_date = datetime.strptime(str(due_str).strip(), "%Y-%m-%d").date()
            paid_date = datetime.strptime(str(paid_str).strip(), "%Y-%m-%d").date()

            if paid_date > due_date:
                days_late = (paid_date - due_date).days
                late_students.append((name, days_late))

            if amount is not None:
                total_collected += float(amount)
        else:
            unpaid_students.append(name)

    # 1. Students who paid late
    print("\n1. Students who paid late (and how many days late):")
    if late_students:
        for name, days in late_students:
            print(f"  - {name} ({days} days late)")
    else:
        print("  None")

    # 2. Students who have not paid
    print("\n2. Students who have not paid:")
    if unpaid_students:
        for name in unpaid_students:
            print(f"  - {name}")
    else:
        print("  None")

    # 3. Total amount collected
    print(f"\n3. Total amount collected: {total_collected:,.2f} USD")
```
